#  Feature Engineering & Data Analysis

```python
date_columns = [
    "order_datetime",
    "promised_datetime",
    "actual_delivery_datetime"
]

for col in date_columns:
    fact_shipments[col] = pd.to_datetime(
        fact_shipments[col],
        errors="coerce"
    )
```

# delivery performance measures

```python
print(fact_shipments[["order_datetime","actual_delivery_datetime"]])
```

```python
# Then calculate delivery time:


fact_shipments["delivery_time_minuts"] = (
    fact_shipments["actual_delivery_datetime"]
    - fact_shipments["order_datetime"]
).dt.total_seconds() / 60
```

```python
fact_shipments["delivery_time_minuts"]
```

```python
# Calculate whether delivery was late:
fact_shipments["is_late"] = (
    fact_shipments["actual_delivery_datetime"]
    > fact_shipments["promised_datetime"]
)
```

```python
# Delivered flag
fact_shipments["is_delivered"] = (
    fact_shipments["actual_delivery_datetime"].notna()
)
```

```python
fact_shipments["is_late"]
```

```python
# Calculate delay:
fact_shipments["delay_minutes"] = (
    fact_shipments["actual_delivery_datetime"]
    - fact_shipments["promised_datetime"]
).dt.total_seconds() / 60
```

```python
fact_shipments["delay_minutes"] 
```

```python
fact_shipments.loc[
    fact_shipments["actual_delivery_datetime"].isna(),
    "delay_minutes"
] = pd.NA
```

# Top management KPIs

```python
total_shipments = len(fact_shipments)

delivered_shipments = fact_shipments["is_delivered"].sum()

undelivered_shipments = (
    total_shipments - delivered_shipments
)

total_revenue = fact_shipments["delivery_revenue"].sum()

avg_revenue_per_shipment = (
    total_revenue / total_shipments
    if total_shipments > 0 else 0
)

delivery_success_rate = (
    delivered_shipments / total_shipments * 100
    if total_shipments > 0 else 0
)

delivered_df = fact_shipments[
    fact_shipments["is_delivered"]
]

on_time_shipments = (
    ~delivered_df["is_late"]
).sum()

on_time_rate = (
    on_time_shipments / len(delivered_df) * 100
    if len(delivered_df) > 0 else 0
)

late_shipments = (
    delivered_df["is_late"].sum()
)

late_rate = (
    late_shipments / len(delivered_df) * 100
    if len(delivered_df) > 0 else 0
)

avg_weight = fact_shipments["weight_kg"].mean()

avg_distance = fact_shipments["distance_km_x"].mean()

avg_delay = delivered_df["delay_minutes"].mean()
```

```python
kpis = {
    "Total Shipments": f"{total_shipments:,}",
    "Delivered": f"{delivered_shipments:,}",
    "Undelivered": f"{undelivered_shipments:,}",
    "Total Revenue": f"{total_revenue:,.2f}",
    "Revenue / Shipment": f"{avg_revenue_per_shipment:,.2f}",
    "Delivery Success": f"{delivery_success_rate:.1f}%",
    "On-Time Delivery": f"{on_time_rate:.1f}%",
    "Late Delivery": f"{late_rate:.1f}%",
  "Avg Weight": f"{avg_weight:,.1f} kg",
    "Avg Distance": f"{avg_distance:,.1f} km",
    "Avg Delay": f"{avg_delay:,.1f} min"
}



kpis
```

```python
late_by_service = (
    delivered_df
    .groupby("service_type")
    .agg(
        delivered=("shipment_id", "count"),
        late_shipments=("is_late", "sum")
    )
    .reset_index()
)

late_by_service["late_rate"] = (
    late_by_service["late_shipments"]
    / late_by_service["delivered"]
    * 100
)

fig = px.bar(
    late_by_service.sort_values("late_rate"),
    x="service_type",
    y="late_rate",
    title="Late Delivery Rate by Service Type",
    text="late_rate"
)

fig.update_traces(
    texttemplate="%{text:.1f}%"
)

fig.show()
```

```python
late_by_branch = (
    delivered_df
    .groupby("branch_name")
    .agg(
        delivered=("shipment_id", "count"),
        late_shipments=("is_late", "sum")
    )
    .reset_index()
)

late_by_branch["late_rate"] = (
    late_by_branch["late_shipments"]
    / late_by_branch["delivered"]
    * 100
)

fig = px.bar(
    late_by_branch.sort_values("late_rate"),
    x="late_rate",
    y="branch_name",
    orientation="h",
    title="Late Delivery Rate by Branch",
    text="late_rate"
)

fig.update_traces(
    texttemplate="%{text:.1f}%"
)

fig.show()
```
